## Завантаження і ознайомлення з даними

```python
import pandas as pd 
import sqlite3
import matplotlib.pyplot as plt
import seaborn as sns
import yaml

with open("./config/configs.yaml", "r") as f:
    config = yaml.safe_load(f)

db_path = config["data"]["db_path"]

# Підключення до БД
conn = sqlite3.connect(db_path)
df = pd.read_sql_query("SELECT * FROM credit_data", conn)
```

```python
print(df.info())
```

```python
display(df.head())
```

```python
display(df.describe())
```

```python
print(f"Number of duplicates: {df.duplicated().sum()}")
print(f"Number of missing values:\n{df.isna().sum()}")
```

**Розподіл даних**

```python
plt.figure(figsize=(10, 6))
sns.countplot(x="default", data=df, palette='viridis', hue="default", legend=False)
plt.title("Distribution of defaults")
plt.ylabel("Number of transactions")
plt.grid(False)
plt.show()
```

## Аналіз збалансованого датасету

```python
conn = sqlite3.connect(db_path)
df_balanced = pd.read_sql_query("SELECT * FROM balanced_credit_data", conn)
```

```python
print(df_balanced.info())
```

```python
display(df_balanced.describe())
```

```python
print(f"Number of duplicates: {df_balanced.duplicated().sum()}")
print(f"Number of missing values:\n{df_balanced.isna().sum()}")
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(x="default", data=df_balanced, palette='viridis', hue="default", legend=False)
plt.title("Distribution of defaults")
plt.ylabel("Number of transactions")
plt.grid(False)
plt.show()
```

**Кореляція ознак**

```python
plt.figure(figsize=(12, 8))
sns.heatmap(df_balanced.corr(), cmap='coolwarm', annot=False)
plt.title('Кореляція ознак')
plt.show()
```

**Виявлення аномалій за допомогою IsolationForest**

```python
from sklearn.ensemble import IsolationForest
# --- Крок 4: Визначення нових аномалій (ML) ---
# Вибираємо лише числові стовпці для моделі
numeric_cols = df_balanced.select_dtypes(include=['number']).columns

# Isolation Forest для виявлення 1% найбільш підозрілих записів
clf = IsolationForest(contamination=0.01, random_state=42)
df_balanced['anomaly'] = clf.fit_predict(df_balanced[numeric_cols])

# -1 — це аномалія, 1 — нормальний запис
anomalies_count = (df_balanced['anomaly'] == -1).sum()
print(f"\nIsolation Forest виявив {anomalies_count} аномалій у новому датасеті.")

# Виведемо приклади знайдених аномалій
display(df_balanced[df_balanced['anomaly'] == -1].head())
```
